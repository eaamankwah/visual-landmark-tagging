<img src="static_images/icons/noun-question-mark-869751.png" alt="?" style="width:25px"/> **Question:** Describe your chosen procedure for preprocessing the data. 
- How does your code resize the images (by cropping, stretching, etc)?  What size did you pick for the input tensor, and why?
- Did you decide to augment the dataset?  If so, how (through translations, flips, rotations, etc)?  If not, why not?

<img src="static_images/icons/noun-answer-3361020.png" alt=">" style="width:25px"/> **Answer**: 

I resize every image so that its shorter side is 256 pixels (`transforms.Resize(256)`), then take a crop of 224x224 (`RandomCrop` for training, `CenterCrop` for validation/test). 224x224 is the standard input size used by most well-known CNN architectures (VGG, ResNet, etc.), it is small enough to keep training fast while still preserving enough detail to recognize a landmark, and using the same crop size for all three splits keeps the model's expected input shape consistent.

I use a `RandomCrop` (instead of a `CenterCrop`) plus `RandomHorizontalFlip`, `RandomRotation(10)` and a mild `ColorJitter` (brightness/contrast/saturation) for the training set. These augmentations simulate the natural variability of user-submitted photos (different angles, lighting conditions, and framing), which helps the model generalize instead of memorizing the exact pixel arrangement of the training images, and reduces overfitting. I deliberately did **not** augment the validation/test transforms, since we want those to reflect the true, unmodified distribution the model will be evaluated (and eventually used) on.

Finally, every split is converted to a tensor and normalized using the mean and standard deviation computed over the dataset itself (via `compute_mean_and_std`), which keeps the pixel values in a well-behaved range for training (roughly zero mean, unit variance per channel).

<img src="static_images/icons/noun-question-mark-869751.png" alt="?" style="width:25px"/> **Question**: Outline the steps you took to get to your final CNN architecture and your reasoning at each step.  

<img src="static_images/icons/noun-answer-3361020.png" alt=">" style="width:25px"/> __Answer:__ 

I started from the classic VGG-style pattern taught in the course: stack blocks of `Conv2d -> BatchNorm2d -> ReLU -> MaxPool2d`, doubling the number of channels while halving the spatial resolution at every block. This is a well-tested pattern that lets the network build up increasingly abstract, higher-level features (edges -> textures -> parts -> whole landmarks) while keeping the computational cost manageable.

Concretely, I used 5 convolutional blocks, going from 3 input channels (RGB) to 16, 32, 64, 128 and finally 256 channels. Starting at 224x224, after 5 rounds of 2x2 max pooling the feature map is down to 7x7, which keeps the flattened feature vector (256 * 7 * 7) reasonably sized for the fully-connected head.

- I added a `BatchNorm2d` after every convolution: this stabilizes and speeds up training, and lets me use a higher learning rate without the training diverging.
- I used `MaxPool2d` rather than a strided convolution for downsampling, since it's simple, parameter-free, and works well in from-scratch CNNs of this size.
- For the classifier head I used one hidden fully-connected layer (512 units) with `BatchNorm1d`, `ReLU` and `Dropout(p=dropout)` before the final linear layer that outputs `num_classes` logits. The dropout layer is the main regularizer fighting overfitting, since with only 50 landmark classes and a modest amount of training data per class, a CNN from scratch can overfit quickly.
- The final layer intentionally has no activation function (no softmax) since `nn.CrossEntropyLoss` expects raw logits and applies `log_softmax` internally; softmax is only applied later, at inference time, in `src/predictor.py`.

This architecture is a good trade-off for a from-scratch model: it's deep enough to learn meaningful hierarchical features for a 50-class landmark classification problem, but still shallow/light enough to train from random initialization in a reasonable amount of time on a single GPU.

Alright, good job getting here! Now it's time to see if all our hard work pays off. In the following cell we will train your model and validate it against the validation set.

Let's start by defining a few hyperparameters. Feel free to experiment with different values and try to optimize your model:

````python
batch_size = 64  # size of the minibatch for stochastic gradient descent (or Adam)
valid_size = 0.2  # fraction of the training data to reserve for validation
num_epochs = 50  # number of epochs for training
num_classes = 50  # number of classes. Do not change this
dropout = 0.5     # dropout for our model
learning_rate = 0.001  # Learning rate for SGD (or Adam)
opt = 'adam'      # optimizer. 'sgd' or 'adam'
weight_decay = 0.01 # regularization. Increase this to combat overfitting
````

````python
from src.data import get_data_loaders
from src.train import optimize
from src.optimization import get_optimizer, get_loss
from src.model import MyModel

# train the model
data_loaders = get_data_loaders(batch_size=batch_size)
model = MyModel(num_classes=num_classes, dropout=dropout)

optimizer = get_optimizer(
    model, learning_rate=learning_rate, optimizer=opt, weight_decay=weight_decay
)
loss = get_loss()

optimize(
    data_loaders,
    model,
    optimizer,
    loss,
    n_epochs=num_epochs,
    save_path="checkpoints/best_val_loss.pt",
    interactive_tracking=True
)
````

Before moving on, let's check the **train** and **validation** accuracy of the model we just trained. This lets us compare different hyperparameter configurations (and spot overfitting) using only train/validation data, without looking at the test set.

````python
import torch
from src.train import one_epoch_test

# (Optional) load the checkpoint with the best validation loss, in case training
# continued for a while after the best epoch
model.load_state_dict(torch.load('checkpoints/best_val_loss.pt'))

print("Training set performance:")
one_epoch_test(data_loaders['train'], model, loss)

print("\nValidation set performance:")
one_epoch_test(data_loaders['valid'], model, loss)
````
